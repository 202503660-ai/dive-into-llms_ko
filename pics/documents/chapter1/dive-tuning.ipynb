### 1. 사전 준비:
#### 1.1 툴킷 알아보기: Transformers
https://github.com/huggingface/transformers

> 🤗 Transformers는 최신 사전학습 모델을 쉽게 다운로드하고 학습할 수 있는 API와 도구를 제공합니다. 사전학습 모델을 사용하면 연산량과 탄소 배출을 줄일 수 있으며, 처음부터 학습하는 데 필요한 시간과 리소스를 절약할 수 있습니다. 이러한 모델들은 다양한 모달리티의 일반적인 작업들을 지원합니다:
📝 자연어 처리: 텍스트 분류, 개체명 인식, 질의응답, 언어 모델링, 요약, 번역, 객관식 선택 및 텍스트 생성.
🖼️ 컴퓨터 비전: 이미지 분류, 객체 탐지 및 시맨틱 세그멘테이션.
🗣️ 오디오: 자동 음성 인식 및 오디오 분류.
🐙 멀티모달: 테이블 질의응답, 광학 문자 인식(OCR), 스캔 문서 정보 추출, 비디오 분류 및 시각적 질의응답(VQA).

상세 문서: https://huggingface.co/docs/transformers/index (영문 공식 문서: https://huggingface.co/docs/transformers/main/en/index)

> 기본 PyPI 환경에서 바로 설치할 수 있습니다: `pip install [Packages]` (네트워크 환경에 따라 필요시 국내외 미러 서버를 지정할 수 있습니다).

- 5. `conda install pytorch`

> PyTorch 공식 설치 가이드: https://pytorch.org/get-started/locally/ (Conda 채널 관리: https://docs.conda.io/projects/conda/en/latest/user-guide/tasks/manage-channels.html)

#### 3.2 모델 학습
검증 세트(dev)에서 검증하고 테스트 세트(test)에서 예측을 동시에 수행하려면 아래 스크립트를 실행합니다:

```shell
python run_classification.py \
    --model_name_or_path  bert-base-uncased \
    --train_file data/train.csv \
    --validation_file data/val.csv \
    --test_file data/test.csv \
    --shuffle_train_dataset \
    --metric_name accuracy \
    --text_column_name "text" \
    --text_column_delimiter "\n" \
    --label_column_name "target" \
    --do_train \
    --do_eval \
    --do_predict \
    --max_seq_length 512 \
    --per_device_train_batch_size 32 \
    --learning_rate 2e-5 \
    --num_train_epochs 1 \
    --output_dir experiments/
```

오류가 발생하거나 멈추는 경우 대개 네트워크 문제입니다:

1. <u>모델 다운로드 시 "Network is unreachable"이 표시되면 모델을 로컬에 수동으로 다운로드합니다: https://huggingface.co/google-bert/bert-base-uncased </u>
2. <u>데이터 로드 후 멈춰서 진행되지 않고, CTRL+C로 종료했을 때 "connection"에서 멈춘 것으로 표시되면 evaluate 패키지가 평가 지표를 로드할 때 네트워크 연결에 실패했기 때문입니다.</u>

![bug](./assets/9.png)

이 경우 evaluate의 GitHub(https://github.com/huggingface/evaluate/tree/main)에서 전체 패키지를 다운로드하고, 하이퍼파라미터에서 `--metric_name` 경로를 로컬 지표 경로로 변경합니다:

```shell
python run_classification.py \
    --model_name_or_path  bert-base-uncased \
    --train_file data/train.csv \
    --validation_file data/val.csv \
    --test_file data/test.csv \
    --shuffle_train_dataset \
    --metric_name evaluate/metrics/accuracy/accuracy.py \
    --text_column_name "text" \
    --text_column_delimiter "\n" \
    --label_column_name "target" \
    --do_train \
    --do_eval \
    --do_predict \
    --max_seq_length 512 \
    --per_device_train_batch_size 32 \
    --learning_rate 2e-5 \
    --num_train_epochs 1 \
    --output_dir experiments/
```
